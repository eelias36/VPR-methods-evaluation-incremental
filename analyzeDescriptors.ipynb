```python
# for all images in a directory, convert them to 'L' mode and save them to a new directory
import os
input_dir = '/home/eelias36/repos/RSO-place-recognition/data/renders/tir/'
output_dir = '/home/eelias36/repos/VPR-methods-evaluation-incremental/assets/spacecraft/tir_8bit/'
os.makedirs(output_dir, exist_ok=True)
for filename in os.listdir(input_dir):
    if filename.endswith('.png'):
        new_filename = str(int(filename[:-4]) // 10000000000) + '.png'
        image = Image.open(os.path.join(input_dir, filename))
        gray_image = image.convert('RGB')
        gray_image.save(os.path.join(output_dir, new_filename))
```

```python
# copy all files from one directory to another and divide the filename by 1e9
import shutil
import os
input_dir = '/home/eelias36/repos/RSO-place-recognition/data/renders/tir_8bit/'
output_dir = '/home/eelias36/repos/VPR-methods-evaluation-incremental/assets/spacecraft/tir_8bit/'
os.makedirs(output_dir, exist_ok=True)
for filename in os.listdir(input_dir):
    if filename.endswith('.png'):
        new_filename = str(int(filename[:-4]) // 10000000000) + '.png'
        shutil.copy(os.path.join(input_dir, filename), os.path.join(output_dir, new_filename))
```
